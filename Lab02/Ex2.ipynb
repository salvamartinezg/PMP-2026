# Laboratory 2: Bayes' Formula

## Step 1: Understanding the Problem

In this laboratory, we will apply Bayes' theorem to a medical diagnostic test.

We want to determine the probability that a person actually has a disease given that their diagnostic test result is positive.

The problem provides the following information:

- The prevalence of the disease is 1% of the population.
- The sensitivity of the test is 95%.
- The specificity of the test is 90%.

We will solve the exercise in three parts:

- **Part (a):** Calculate the theoretical probability of having the disease given a positive test result and explain the result.
- **Part (b):** Simulate the problem in Python and compare the experimental probability with the theoretical result.
- **Part (c):** Calculate the minimum specificity required for the probability of having the disease after a positive test to reach 50%.

We will use Bayes' theorem to update the initial probability of having the disease after observing a positive test result.

## Step 2: Importing Libraries and Defining the Parameters

In this step, we import the necessary Python libraries and define the parameters of the problem.

We will use NumPy to generate random values and simulate a large population.

The parameters are:

- Disease prevalence: 1%.
- Test sensitivity: 95%.
- Test specificity: 90%.

The specificity is the probability of obtaining a negative test result when the person does not have the disease:

$$
P(-\mid B^c)=0.90
$$

Since the test result can be either positive or negative, we have:

$$
P(+\mid B^c)+P(-\mid B^c)=1
$$

Therefore, the probability of a false positive is:

$$
P(+\mid B^c)=1-P(-\mid B^c)
$$

Substituting the specificity value:

$$
P(+\mid B^c)=1-0.90=0.10
$$

Therefore, 10% of individuals without the disease are expected to receive a positive test result.

```python
import numpy as np

# Set a random seed for reproducibility
rng = np.random.default_rng(42)

# Define the parameters
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90

# Calculate the false positive rate
false_positive_rate = 1 - specificity

# Number of individuals in the simulation
N = 1_000_000

print(f"Disease prevalence: {prevalence:.2%}")
print(f"Test sensitivity: {sensitivity:.2%}")
print(f"Test specificity: {specificity:.2%}")
print(f"False positive rate: {false_positive_rate:.2%}")
print(f"Population size: {N:,}")
```

## Step 3: Calculating the Theoretical Probability

We want to calculate the probability that a person has the disease given that their test result is positive.

Using Bayes' theorem:

$$
P(B \mid +)=\frac{P(+ \mid B)P(B)}{P(+)}
$$

The probability of a positive test can be calculated using the law of total probability:

$$
P(+)=P(+ \mid B)P(B)+P(+ \mid \neg B)P(\neg B)
$$

Therefore, Bayes' theorem can be written as:

$$
P(B \mid +)=
\frac{P(+ \mid B)P(B)}
{P(+ \mid B)P(B)+P(+ \mid \neg B)P(\neg B)}
$$

The known values are:

- $P(B)=0.01$
- $P(\neg B)=0.99$
- $P(+ \mid B)=0.95$
- $P(+ \mid \neg B)=0.10$

We substitute these values into the formula to calculate the theoretical probability.

```python
# Calculate the probability of not having the disease
p_no_disease = 1 - prevalence

# Calculate the probability of a positive test
p_positive = (
    sensitivity * prevalence
    + false_positive_rate * p_no_disease
)

# Apply Bayes' theorem
p_disease_given_positive = (
    sensitivity * prevalence / p_positive
)

print(f"Probability of a positive test: {p_positive:.6f}")

print(
    "Theoretical probability of disease "
    f"given a positive test: {p_disease_given_positive:.6f}"
)

print(
    "Theoretical probability in percentage: "
    f"{p_disease_given_positive:.2%}"
)
```

## Step 4: Interpreting the Theoretical Result

The theoretical probability of having the disease given a positive test result is approximately 8.76%.

Although the test has a sensitivity of 95% and a specificity of 90%, a positive result does not mean that the person is very likely to have the disease.

This happens because the disease is rare and affects only 1% of the population.

Consequently, even though the test detects most people who have the disease, the number of people without the disease is much larger. Some of these people will receive false-positive results.

Therefore, the initial probability of having the disease must be considered when interpreting a positive test result.

**Conclusion:** A person with a positive test result has approximately an 8.76% probability of actually having the disease under the assumptions of this model.

## Step 5: Simulating the Experiment

In this step, we simulate a population of one million individuals using NumPy.

For each individual, we generate two outcomes:

1. Disease status: the individual has a 1% probability of having the disease.
2. Test result: the probability of a positive result depends on the individual's disease status.

If the individual has the disease, the probability of a positive test is 95%.

If the individual does not have the disease, the probability of a positive test is 10%.

After simulating the population, we select all individuals with positive test results and calculate the proportion who actually have the disease.

This proportion provides an experimental estimate of the probability calculated theoretically in Step 3.

```python
# Simulate disease status for the entire population
disease = rng.random(N) < prevalence

# Identify individuals with and without the disease
has_disease = disease
has_no_disease = ~disease

# Initialize the test results
test_positive = np.zeros(N, dtype=bool)

# Simulate test results for individuals with the disease
test_positive[has_disease] = (
    rng.random(has_disease.sum()) < sensitivity
)

# Simulate test results for individuals without the disease
test_positive[has_no_disease] = (
    rng.random(has_no_disease.sum()) < false_positive_rate
)

# Count positive tests
number_positive = test_positive.sum()

# Count true positives
number_true_positive = np.sum(disease & test_positive)

# Estimate the probability of disease given a positive test
p_experimental = number_true_positive / number_positive

print(f"Total population: {N:,}")
print(f"Individuals with the disease: {disease.sum():,}")
print(f"Positive test results: {number_positive:,}")
print(f"True positive results: {number_true_positive:,}")

print(
    f"Experimental probability: {p_experimental:.6f}"
)

print(
    f"Experimental probability: {p_experimental:.2%}"
)
```

## Step 6: Comparing Experimental and Theoretical Results

In this step, we compare the experimental probability obtained from the simulation with the theoretical probability calculated using Bayes' theorem.

The absolute error is calculated as:

$$
\text{Absolute Error} =
\left|P_{\text{experimental}}-P_{\text{theoretical}}\right|
$$

A small absolute error indicates that the experimental estimate is close to the theoretical result.

The two probabilities may differ because the simulated population is finite and the outcomes are random.

Increasing the number of simulated individuals generally reduces the sampling variability and improves the precision of the estimate.

```python
# Calculate the absolute error
absolute_error = abs(
    p_experimental - p_disease_given_positive
)

print(
    f"Theoretical probability: "
    f"{p_disease_given_positive:.6f} "
    f"({p_disease_given_positive:.2%})"
)

print(
    f"Experimental probability: "
    f"{p_experimental:.6f} "
    f"({p_experimental:.2%})"
)

print(f"Absolute error: {absolute_error:.6f}")
print(f"Absolute error in percentage points: {absolute_error * 100:.4f}")
```

### Interpretation of the Results

The theoretical probability of having the disease given a positive test result is 8.76%, while the experimental probability obtained from the simulation is 8.84%.

The absolute error is 0.000863, which corresponds to approximately 0.0863 percentage points. This indicates that the experimental probability is very close to the theoretical probability.

The small difference between the two values is due to the random variability of the simulation. Since we simulated one million individuals, the experimental result provides a good approximation of the theoretical probability calculated using Bayes' theorem.

**Conclusion:** The simulation successfully reproduces the theoretical result, demonstrating that Python can be used to estimate probabilities and verify mathematical calculations.

## Step 7: Calculating the Minimum Specificity

**Exercise part (c)**

In this step, we calculate the minimum specificity required for the probability of having the disease given a positive test result to reach 50%.

We keep the disease prevalence at 1% and the test sensitivity at 95%.

Let $s$ represent the specificity. By definition:

$$
s=P(-\mid B^c)
$$

Since the test result can be either positive or negative, we have:

$$
P(+\mid B^c)+P(-\mid B^c)=1
$$

Therefore, the probability of a false positive is:

$$
P(+\mid B^c)=1-s
$$

Using Bayes' theorem:

$$
P(B\mid +)=
\frac{P(+\mid B)P(B)}
{P(+\mid B)P(B)+P(+\mid B^c)P(B^c)}
$$

Substituting $P(+\mid B^c)=1-s$ into the formula:

$$
P(B\mid +)=
\frac{P(+\mid B)P(B)}
{P(+\mid B)P(B)+(1-s)P(B^c)}
$$

We want the posterior probability to equal 50%. Therefore:

$$
0.50=
\frac{0.95(0.01)}
{0.95(0.01)+(1-s)(0.99)}
$$

Solving this equation gives:

$$
1-s=\frac{0.95(0.01)}{0.99}
$$

Therefore:

$$
s=1-\frac{0.95(0.01)}{0.99}
$$

The minimum specificity is approximately 99.04%.


```python
# Desired posterior probability
target_probability = 0.50

# Calculate the minimum specificity
specificity_min = 1 - (
    sensitivity * prevalence * (1 - target_probability)
    / (target_probability * (1 - prevalence))
)

# Calculate the corresponding false positive rate
new_false_positive_rate = 1 - specificity_min


print(f"Minimum specificity: {specificity_min:.6f}")
print(f"Minimum specificity: {specificity_min:.2%}")
print(f"False positive rate: {new_false_positive_rate:.4%}")
```

### Interpretation of the Results

The minimum specificity required for the probability of having the disease given a positive test result to reach 50% is approximately **99.04%**.

This means that the test must correctly identify at least 99.04% of individuals who do not have the disease.

The corresponding false positive rate is approximately **0.9596%**. Therefore, fewer than 1% of individuals without the disease would receive a positive test result.
