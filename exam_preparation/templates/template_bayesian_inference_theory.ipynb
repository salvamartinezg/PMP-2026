# Bayesian Inference: theory questions and answers

Exam-style theoretical questions on Bayesian inference, each followed by a short conceptual
answer ready to paste into a text cell.

### 1. What is Bayesian inference?

Bayesian inference is a way of learning about unknown quantities (parameters) by combining what
we believed before seeing the data with the information contained in the data. The result is not
a single number but a probability distribution over the possible values of the parameter, which
describes both the most plausible values and how uncertain we still are.

### 2. What is the difference between descriptive and inferential statistics?

Descriptive statistics only summarise the data we observed, for example with a mean, a variance
or a plot. Inferential statistics go further and use those data to draw conclusions about the
process that generated them: its unknown parameters and the observations we have not seen yet.

### 3. What do we mean by data, model and parameter?

The data are what we actually observe. The model is the probabilistic mechanism we assume
generated those data. The parameter is the unknown quantity inside the model that we want to
learn. A model is a simplification of reality: it keeps the relevant features and ignores details.

### 4. What does the i.i.d. assumption mean, and when can it fail?

i.i.d. means independent and identically distributed: each observation comes from the same
distribution and does not influence the others. It does not mean the observations are equal. It
fails when the data have a trend, seasonality or memory, for example a time series where today's
value depends on yesterday's. In that case a model that treats the observations as i.i.d. is wrong.

### 5. What are the four ingredients of Bayes' theorem?

The prior is what we believe about the parameter before seeing the data. The likelihood measures
how compatible the observed data are with each possible value of the parameter. The posterior is
our updated belief after seeing the data. The evidence (or marginal likelihood) is the
normalising constant that makes the posterior a proper probability distribution.

$$p(\theta\mid y) = \frac{p(y\mid\theta)\,p(\theta)}{p(y)}$$

### 6. What is the role of the prior distribution?

The prior encodes the information we have about the parameter before observing the data, such as
expert knowledge or results from previous studies. It can be weak (vague), letting the data
dominate, or strong, pulling the conclusions towards what we already believed. Its influence is
largest when there are few data and fades as the amount of data grows.

### 7. What is the likelihood, and is it a probability distribution over the parameter?

The likelihood is the probability of the observed data seen as a function of the parameter. It
tells us which parameter values make the data more or less plausible. It is not a probability
distribution over the parameter: it does not have to integrate to one over the parameter values.

### 8. What is the posterior distribution?

The posterior is the distribution of the parameter after taking the data into account. It is
proportional to the likelihood times the prior, so it combines both sources of information. All
Bayesian conclusions, such as point estimates, intervals and predictions, are derived from it.

### 9. Why can we often ignore the denominator of Bayes' theorem?

The evidence does not depend on the parameter, so it only rescales the posterior without changing
its shape. That is why we usually write that the posterior is proportional to the likelihood times
the prior, and normalise at the end if needed.

$$p(\theta\mid y)\propto p(y\mid\theta)\,p(\theta)$$

### 10. What is the difference between p(y | θ) and p(θ | y)?

The first is the likelihood: the probability of the data for a fixed value of the parameter. The
second is the posterior: the probability of the parameter given the data we observed. Confusing
them is the "reversed conditional" error, like confusing the probability of the evidence given
guilt with the probability of guilt given the evidence. To go from one to the other we need the prior.

### 11. Why does a very accurate medical test not give a very high probability of disease after a positive result?

Because the result depends not only on the accuracy of the test but also on the prior, that is,
on how common the disease is. If the disease is rare, the small fraction of false positives among
the many healthy people can outnumber the true positives among the few sick people, so most
positive results are false alarms.

### 12. What is a conjugate prior and why is it useful?

A prior is conjugate to a likelihood when the posterior belongs to the same family of
distributions as the prior. This is useful because the posterior can be obtained exactly by simply
updating the parameters of the prior with the data, without integrals or simulation.

### 13. Which conjugate pairs do we use, and what is the update rule?

For a probability of success with binomial data we use a Beta prior; the posterior is again a
Beta, where the first parameter grows with the number of successes and the second with the number
of failures. For a rate with Poisson count data we use a Gamma prior; the posterior is again a
Gamma, where the shape grows with the total count and the rate grows with the number of observations.

### 14. How should the parameters of a Beta prior be interpreted?

They can be read as pseudo-counts: roughly, the number of successes and failures we "saw" before
the experiment. Their ratio sets where the prior is centred and their sum sets how strong it is.
Beta(1, 1) is uniform and expresses no preference; Beta(20, 20) is concentrated around one half;
Beta(1, 4) favours small values.

### 15. How do we choose the parameters of a Gamma prior for a rate?

We choose them so that the prior mean matches what we expect before seeing the data, and so that
its strength reflects how much we trust that expectation. The rate parameter behaves like the
number of intervals already observed, so a small value gives a weak prior that the data quickly
override, while a large value gives a strong prior based on reliable historical information.

### 16. What happens to the posterior when we collect more data?

The posterior usually becomes narrower, which means less uncertainty about the parameter, and it
concentrates around the value suggested by the data. The influence of the prior decreases, so
analysts who started with different reasonable priors end up with similar posteriors.

### 17. What are the MLE, the MAP estimate and the posterior mean?

The maximum likelihood estimate (MLE) is the parameter value that makes the observed data most
probable; it uses only the data. The maximum a posteriori (MAP) estimate is the most probable
value of the posterior, that is, its mode. The posterior mean is the average value of the
parameter under the posterior.

### 18. When do the MLE and the MAP estimate coincide?

They coincide when the prior is flat (uniform), because multiplying the likelihood by a constant
does not change where its maximum is. With an informative prior the MAP is pulled away from the
MLE towards the values the prior favours.

### 19. Why is the posterior mean usually between the prior mean and the MLE?

Because the posterior combines both sources of information. The posterior mean is a compromise:
it is pulled towards the prior mean by the prior and towards the MLE by the data. The more data we
have, the closer it gets to the MLE.

### 20. Why is a single point estimate not enough?

A point estimate says nothing about how certain we are. The same estimate can come from very few
or from very many observations, which deserve very different levels of confidence. That is why we
report the whole posterior, or at least a measure of spread and a credible interval.

### 21. What is a credible interval and how is it interpreted?

A credible interval is a range of parameter values that contains a given posterior probability,
for example 94 %. It is interpreted directly: given the model and the data, there is a 94 %
probability that the parameter lies in that interval. This is different from a frequentist
confidence interval, which is a statement about the procedure, not about the parameter.

### 22. What is the HDI and how does it differ from an equal-tailed interval?

The highest density interval (HDI) contains the most probable values of the parameter: every
point inside it has higher posterior density than any point outside, and for unimodal
distributions it is the shortest interval with the required probability. An equal-tailed interval
instead leaves the same probability in each tail. Both coincide for symmetric posteriors but
differ for skewed ones, and for multimodal posteriors the HDI may consist of several separate pieces.

### 23. What is the posterior predictive distribution?

It is the distribution of a future observation given the data we have already seen. It averages
the predictions of the model over all plausible parameter values, weighted by the posterior, so it
accounts for our uncertainty about the parameter. For that reason it is wider than a prediction
made by plugging in a single estimate.

### 24. How can the posterior predictive distribution be obtained by simulation?

First draw a parameter value from the posterior, then simulate a new observation from the model
using that value, and repeat many times. The simulated observations follow the posterior
predictive distribution.

### 25. If the predicted probability of heads for the next toss is 0.7, will exactly 7 of the next 10 tosses be heads?

No. The value 0.7 is a probability for each individual toss, not a guaranteed count. The number of
heads in the next ten tosses is still random; it will often be around seven, but other values are
also possible.

### 26. What is a posterior predictive check?

It is a way of checking the model: we simulate replicated data sets from the posterior predictive
distribution and compare them with the data we actually observed. If the model cannot reproduce
important features of the real data, such as their spread or extreme values, the model is
inadequate and must be revised.

### 27. What can go wrong in a Bayesian analysis?

The prior may be too restrictive or inappropriate, the i.i.d. assumption may be violated, or the
likelihood may ignore important mechanisms of the real process. Moreover, a narrow posterior does
not guarantee that the model is correct: Bayesian inference measures uncertainty only within the
chosen model, so the model must still be checked against the data.

### 28. What is the general Bayesian workflow?

We start from data and a model with a prior, combine them into a posterior, use the posterior to
summarise the parameters and to make predictions, and finally check whether the model reproduces
the data. If it does not, we revise the model and repeat.
