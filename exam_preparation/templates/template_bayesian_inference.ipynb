```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import arviz as az

def hdi(samples, prob=0.94):
    """Works with ArviZ 0.x (hdi_prob) and 1.x (prob)."""
    try:
        return az.hdi(samples, hdi_prob=prob)
    except TypeError:
        return az.hdi(samples, prob=prob)
```

```python
# Beta-Binomial: y successes in N trials, prior Beta(a, b)
a, b, N, y = 1, 1, 10, 8
post = stats.beta(a + y, b + N - y)
print("MAP =", (a + y - 1) / (a + b + N - 2), "  mean =", post.mean(), "  94% HDI =", hdi(post.rvs(100_000, random_state=0)))
```

```python
# Gamma-Poisson: total count S over n intervals, prior Gamma(alpha, beta) with beta = RATE
alpha, beta, S, n = 2, 0.1, 180, 10
post = stats.gamma(a=alpha + S, scale=1 / (beta + n))     # scipy: scale = 1 / rate
samples = post.rvs(100_000, random_state=0)
print("mode =", (alpha + S - 1) / (beta + n), "  mean =", post.mean(), "  94% HDI =", hdi(samples))

# Colab / ArviZ 0.x one-liner:  az.plot_posterior(samples, hdi_prob=0.94, point_estimate="mode")
x = np.linspace(post.ppf(0.001), post.ppf(0.999), 400)
plt.plot(x, post.pdf(x))
plt.title("posterior")
plt.savefig("posterior.png", dpi=150, bbox_inches="tight")
plt.show()
```
